# Multi-agent research report

A planner designs focused searches and an outline. Search tools gather source excerpts; a research agent writes an evidence brief; a writer drafts the report; a reviewer critiques it; and an editor revises it. Citations are linked to retrieved source records.

This notebook is a standalone adaptation of a course assignment. See `README.md` for setup and limitations.


## 1. Load the workflow

Create a local `.env` from `.env.example` before running the report cell. Keep `.env` out of Git.


```python
from pathlib import Path
import re
from dotenv import load_dotenv
from IPython.display import Markdown, display
from research_workflow import ReportConfig, run_workflow

load_dotenv(Path.cwd() / ".env")
```

## 2. Choose a topic and run the agents

The default target is about 1,200 words. Change the topic or `ReportConfig` to suit your report. The workflow makes several model calls and search requests.


```python
topic = "The ensemble Kalman filter for time series forecasting"
config = ReportConfig(target_words=1200, minimum_words=900)
result = run_workflow(topic, config=config)

print("Search queries:")
for query in result["plan"]["queries"]:
    print(f"- {query}")
print("Quality checks:", result["quality"])
display(Markdown(result["report"]))
```

## 3. Save the report

Generated reports go to `reports/`, which is ignored by Git. Review the source links and factual claims before sharing a report.


```python
report_dir = Path("reports")
report_dir.mkdir(exist_ok=True)
slug = re.sub(r"[^a-z0-9]+", "-", topic.lower()).strip("-")[:60]
report_path = report_dir / f"{slug}.md"
report_path.write_text(result["report"], encoding="utf-8")
print(f"Saved to {report_path.resolve()}")
```
